```python
from abc import ABC, abstractmethod


class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class CreditCardPayment(Payment):
    def pay(self, amount):
        print("Paid ₹", amount, "using Credit Card")


class UPIPayment(Payment):
    def pay(self, amount):
        print("Paid ₹", amount, "using UPI")


payment1 = CreditCardPayment()
payment2 = UPIPayment()

payment1.pay(1500)
payment2.pay(800)
```
