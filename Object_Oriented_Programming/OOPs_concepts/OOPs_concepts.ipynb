Defining a class without self keyword

```python
class Car:

    name : str
    color : str
    car_type : str
```

```python
kia = Car()

kia.name = "Kia"
kia.color = "White"
kia.car_type = "Petrol + CNG"
```

```python
def intro(self):
        print(f"Name of car is : {self.name}")
        print(f"Color of car : {self.color}")
        print(f"Type of car : {self.car_type}")
```

```python
intro(kia)
```

Types of Methods : 

    1) Instance Method
    2) Static Method
    3) Class Method

Instance Method

    Methods / functions defined in class will give different outputs according to objects or w.r.t to object 

```python
class Avenger:

    def __init__(self , name , weapon):
        self.name = name
        self.weapon = weapon

    def intro(self) :
        print(f"Name of Avenger is {self.name}")
        print(f"Weapon of {self.name} is {self.weapon}\n")

iron_man = Avenger("Tony Stark" , "Mark_42")

iron_man.intro()

captain = Avenger("Steve Rogers" , "Sheild")
captain.intro()
```

Intro function is changing outputs w.r.t Objects

Class Method

    Functions are defined in such a way that it will remain same for every object

```python
class Avengerr :

    origin = "One of the Avengers"

    def __init__(self , name , weapon):
        self.name = name
        self.weapon = weapon

    @classmethod
    def origin_info(cls,self):
        print(f"Origin of {self.name} is {cls.origin}")

    def self_intro(self):
        print(f"Name of Avenger is {self.name}")
        print(f"Weapon of {self.name} is {self.weapon}\n")
```

```python
iron_man = Avengerr("Tony_Stark" , "Mark_42")
iron_man.origin_info(iron_man)
iron_man.self_intro()

captain = Avengerr("Steve Rogers" , "Sheild")
captain.origin_info(captain)
captain.self_intro()
```

Static Method 

    It doesn't use cls or self

```python
class Avengerrr :

    origin = "One of the Avengers"

    @staticmethod
    def mission():
        print("Mission : To defeat Thanos ..!")

    def __init__(self , name , weapon):
        self.name = name
        self.weapon = weapon

    @classmethod
    def origin_info(cls,self):
        print(f"Origin of {self.name} is {cls.origin}")

    def self_intro(self):
        print(f"Name of Avenger is {self.name}")
        print(f"Weapon of {self.name} is {self.weapon}\n")
```

```python
iron_man = Avengerrr("Tony_Stark" , "Mark_42")
iron_man.origin_info(iron_man)
iron_man.mission()
iron_man.self_intro()

captain = Avengerrr("Steve Rogers" , "Sheild")
captain.origin_info(captain)
captain.mission()
captain.self_intro()
```
