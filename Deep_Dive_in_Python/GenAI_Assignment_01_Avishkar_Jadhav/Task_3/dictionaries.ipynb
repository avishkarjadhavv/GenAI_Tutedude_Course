```python
price_dict = {
    "book": 50,
    "pen": 20,
    "pencil": 10,
    "crayons": 40,
    "rubber": 5,
    "paper": 30
}

print("Price Dictionary:", price_dict)
```

```python
price_dict["notebook"] = 60

price_dict["pen"] = 25

if "rubber" in price_dict:
    del price_dict["rubber"]
else:
    print("Product not found")

if "eraser" in price_dict:
    del price_dict["eraser"]
else:
    print("Eraser not found")

print("Updated Price Dictionary:", price_dict)
```

```python
total = sum(price_dict.values())
count = len(price_dict)

average_price = total / count

print("Average price:", average_price)
```
