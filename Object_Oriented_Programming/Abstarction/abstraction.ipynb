Abstraction :

    Abstraction is hiding the implementation details and only showing the necessary features of the object

    This helps in reducting programming complexity and efforts

How can we achieve abstraction in python :

    1) Abstract Class --> A class that cannot be instantiated or that means for an abstract class we can't create the object of this class

    2) Abstract Method --> Abstract method just because it can't accessed by the object of the abstract class itself thats why we have to call this abstract method in our child class and the implentation of the abstract method is not provided inside the abstract class but must be provided in the child class.

The abstract method is being defined by @abstractmethod 

```python
from abc import ABC , abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vehicle):
    
    def start(self):
        print("Car starts with an ignition key ..")

    def stop(self):
        print("The car stops by break pedals ..")

class ElectricScooter(Vehicle):

    def start(self):
        print("The electric scooter starts with an Power Button ..")

    def stop(self):
        print("The Electric Scooter stops by break pedals ..")
```

```python
car = Car()
car.start()
```

```python
car.stop()
```

```python
elec = ElectricScooter()
elec.start()
```
