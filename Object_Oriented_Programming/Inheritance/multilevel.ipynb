Multilevel Inheritance 

    The properties , Attributes or Methods of the GrandParent Class can be inherited by the parent class and 

    Properties of parent class are inherited by Child class

Therefore , the properties of grandParent class are inherited to child class too

```python
class GrandParent : 

    def __init__(self, home , car , land):
        self.home = home
        self.car = car
        self.land = land

    def GP_intro(self):
        print("GP INTRO -->")
        print(f"The car family owns : {self.car}")
        print(f"The land Family owns : {self.land}\n")

class Parent(GrandParent) : 

    def __init__(self, home , car , land):
        super().__init__(home,car,land)
        self.home = home
        self.car = car
        self.land = land

    def P_intro(self):
        print("P INTRO -->")
        print(f"The car family owns : {self.car}")
        print(f"The land Family owns : {self.land}\n")

class Child(Parent) : 

    def __init__(self, home , car , land):
        super().__init__(home,car,land)
        self.home = home
        self.car = car
        self.land = land

    def C_intro(self):
        print("C INTRO -->")
        print(f"The car family owns : {self.car}")
        print(f"The land Family owns : {self.land}\n")
```

```python
aj = Child("Gajanan Krupa" , "super_splendor" , "No Land")

aj.C_intro()
aj.P_intro()
aj.GP_intro()

print(aj.home)
```
