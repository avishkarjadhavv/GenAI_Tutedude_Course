Polymorphism :
    
    In OOPs , Polymorphism allows objects of different classes as if they are the object of the same class & Its enables reusibility and makes program more flexible


Types of Polymorphism :
        
    1) Method Overriding ( Rum time Polymorphism )
    2) Method Overloading ( Compile Time polymorphism )
    3) Operator Overloading
    

Method Overriding : 

    A child class provides a specific implementation of a method that has already been defined into parent class . This allows dynamic selection at runtime .

```python
class GrandParent:

    def speak(self):
        print("My name is Amit ...")

class Parent(GrandParent):

    def speak(self):
        print("My name is Junior Amit ...")

class Child(Parent):
    # def speak(self):
    #     print("My name is Junior Amit Junior...")
    pass
```

```python
grandpa = GrandParent()
grandpa.speak()
```

```python
pa = Parent()
pa.speak()
```

```python
child = Child()
child.speak()
```

```python
child = Child()
child.speak()
```

Another example :

```python
class Shape:

    def area(self):
        return "Its returning area of a shape ..."

class Rectangle(Shape):

    def __init__(self,l,b):
        self.l = l
        self.b = b

    def area(self):
        return f"Area of Rectangle is {self.l*self.b} ..."

class Circle(Shape):

    def __init__(self,radius):
        self.radius = radius

    def area(self):
        return f"Area of Circle is {3.14 * self.radius * self.radius} ..."
```

```python
shape = Shape()
shape.area()
```

```python
rect = Rectangle(4,2)
rect.area()
```

```python
circle = Circle(7)
circle.area()
```
