Task 2

```python
# Read the entire file using read()

f = open('sales_data.txt', 'r')
content = f.read()
print("1. Entire file:")
print(content)
f.close()


# Read the first line using readline()

f = open('sales_data.txt', 'r')
first_line = f.readline()
print("2. First line:")
print(first_line)
f.close()


# Read all lines using readlines() and convert them to integers

f = open('sales_data.txt', 'r')
lines = f.readlines()
f.close()

sales = []

for line in lines:
    sales.append(int(line.strip()))

print("3. Sales as integers:")
print(sales)
```

Task 4

```python
f = open('sales_data.txt', 'r')

lines = f.readlines()
f.close()

sales = []

for line in lines:
    sales.append(int(line.strip()))

total_sales = sum(sales)
highest_sale = max(sales)
lowest_sale = min(sales)
average_sale = total_sales / len(sales)

print("Total Sales:", total_sales)
print("Highest Sale:", highest_sale)
print("Lowest Sale:", lowest_sale)
print("Average Sale:", average_sale)
```

Task 7

```python
f = open("sales_data.txt", "r")

lines = f.readlines()
f.close()

discounted_sales = []

for line in lines:
    price = int(line.strip())
    discounted_price = price - (price * 10 / 100)
    discounted_sales.append(str(discounted_price) + "\n")

f = open("discounted_sales.txt", "w")
f.writelines(discounted_sales)
f.close()

f = open("discounted_sales.txt", "r")
content = f.read()
print("Discounted Sales:")
print(content)
f.close()
```
