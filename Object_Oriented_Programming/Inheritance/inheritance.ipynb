syntax for Inheritance : 

    class base_class:
        def __init__(self):
            pass

        def intro(self):
            pass

    class derived_class (base_class):
        pass

Now derived_class can access the intro and __init__ function from base_class

Inheritance is a fundamental concept in OOP because of which a child class is able to inherit the properties or Attributes and Functionalities of Parent class

In python , There are 3 types of Inheritance

    1) Single Inheritance
    2) Multilevel Inheritance
    3) Multiple Inheritance

Single Inheritance
    The features or the Attributes of only one parent class will be inherited into one child class

```python
class Avenger:

    def __init__(self , name , weapon):
        self.name = name
        self.weapon = weapon

    def intro(self):
        print(f"My name is {self.name}")
        print(f"My weapon is : {self.weapon}")

class IronMan(Avenger):
    def __init__(self , name , weapon):
        super().__init__(name,weapon)
        self.name = name
        self.weapon = weapon

    def derived_info(self):
        print("This is derived class function")
```

```python
iron = IronMan("Iron Man","Mark_42")

print(iron.name)
print(iron.weapon)
iron.derived_info()
iron.intro()
```
