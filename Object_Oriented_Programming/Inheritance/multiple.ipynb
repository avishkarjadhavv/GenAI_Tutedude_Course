Multiple Inheritance

    The properties of two or more parent class will be inherited to the child class

Syntax for Multiple Inheritance : 

    class child (parent1,parent2):
        pass

```python

class Fighter:
    def fight(self):
        print("Avenger can fight")


class Flyer:
    def fly(self):
        print("Avenger can fly")


# Multiple inheritance
class IronMan(Fighter, Flyer):
    def use_suit(self):
        print("Iron Man uses his suit")


iron_man = IronMan()

iron_man.fight()
iron_man.fly()
iron_man.use_suit()
```
