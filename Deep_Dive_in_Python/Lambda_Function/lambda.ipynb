```python
# Function for square of a number

def square(x):
    return x**2

print(square(3))
```

Syntax for Lambda Funtions :

    lambda argument : expression

* Lambda functions don't have function name .. 

eg. name = lambda argument : expression
    name(argument)

```python
a = lambda x : x**2

a(5)
```

```python
name_person = lambda name : f"My name is {name}"
name_person("Avishkar")
```

```python
name_person("Nilesh")
```
