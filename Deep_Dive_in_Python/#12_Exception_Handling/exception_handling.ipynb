all keywords :

    try --> try block is where we write the code which may give error or exceptions

    except --> except block is used to handle the error in a way such that program execution shouldn't stop

    else --> if try block throw no error , then else block will execute

    finally --> This block is going to run anyway , put this at last

Different type of exceptions : 

    1) ZeroDivisionError --> Dividing by zero
    2) FileNotFoundError --> File not Found
    3) ValueError --> Invalid Value
    4) TypeError --> Invalid Type

```python
a = c
```

```python
try:
    a=b
except:
    print("Value b is not assigned")
```

```python
try:
    a=b
except NameError as exception:
    print(exception)
```

```python
den = int(input("Enter value of denominator : "))

try :
    result = 1 / den
    print(result)

except ZeroDivisionError as ex:
    print(ex)
```

Multiple Exceptions :


```python
b = int(input("Enter the value of denominator b : "))
d = input("Enter the value of denominator c : ")
try:
    result = 1 / b
    print(result)
    a = c
    division = 1/d
    print(division)

except ZeroDivisionError as ex1:
    print(ex1)

except NameError as ex2:
    print(ex2)

except TypeError as ex3:
    print(ex3)
```

When you don't know what exception can come up 

    Use      except Exception:

```python
b = int(input("Enter the value of denominator b : "))
d = input("Enter the value of denominator c : ")
try:
    result = 1 / b
    print(result)
    a = c
    division = 1/d
    print(division)

except Exception as ex1:
    print(ex1)
```

else Block

```python
num = int(input("Enter value of denominator num : "))

try :
    result = 1 / num

except Exception as ex1:
    print(ex1)

else:
    print(f"Result for 1 divide by {num} is {result}")
```

finally Block

```python
num = int(input("Enter value of denominator num : "))

try :
    result = 1 / num

except Exception as ex1:
    print(ex1)

else:
    print(f"Result for 1 divide by {num} is {result}")

finally:
    print("Code execution is completed .. ")
```

FileNotFoundError exception

```python
try:
    f = open("example.txt" , "r")
    content = f.read()
    print(content)

except FileNotFoundError as ex1:
    print(ex1)

finally :
    if 'f' in locals() and not f.closed():
        f.close()
        print("File Closed .. ")
```
