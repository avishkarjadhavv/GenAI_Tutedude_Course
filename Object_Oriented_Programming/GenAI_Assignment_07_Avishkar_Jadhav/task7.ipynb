```python
class Inventory:
    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)
        print(product.name, "added to inventory.")

    def remove_product(self, product_name):
        for product in self.products:
            if product.name == product_name:
                self.products.remove(product)
                print(product_name, "removed from inventory.")
                return

        print("Product not found.")

    def list_products(self):
        print("\nProducts in Inventory:")
        for product in self.products:
            print(product)

    def total_value(self):
        return sum(product.price for product in self.products)


class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.price = price
        self.category = category

    def __str__(self):
        return f"{self.name} - ₹{self.price} ({self.category})"


class Store:
    def __init__(self, name, inventory):
        self.name = name
        self.inventory = inventory

    def display_summary(self):
        print("\nStore:", self.name)
        self.inventory.list_products()
        print("Total inventory value: ₹", self.inventory.total_value())


# Create inventory
inventory = Inventory()

# Create products
product1 = Product("Laptop", 55000, "Electronics")
product2 = Product("Mobile", 25000, "Electronics")
product3 = Product("Notebook", 80, "Stationery")

# Add products
inventory.add_product(product1)
inventory.add_product(product2)
inventory.add_product(product3)

# Create store and display summary
store = Store("AJ's Store", inventory)
store.display_summary()

# Remove a product
inventory.remove_product("Mobile")

# Display updated summary
store.display_summary()
```
