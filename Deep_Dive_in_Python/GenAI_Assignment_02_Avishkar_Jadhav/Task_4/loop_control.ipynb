```python
daily = [200, 150, 0, 400, 50, -1, 300]

total_sales = 0

for sale in daily:

    if sale == -1:
        print("Corrupted data found. Stopping.")
        break

    if sale == 0:
        continue

    total_sales = total_sales + sale
    print("Running total:", total_sales)

print("Total sales:", total_sales)
```

```python
daily = [200, 150, 0, 400, 50, -1, 300]

total_sales = 0
skipped_days = 0

for sale in daily:

    if sale == -1:
        print("Corrupted data found. Stopping.")
        break

    if sale == 0:
        skipped_days = skipped_days + 1
        continue

    total_sales = total_sales + sale

print("Total sales:", total_sales)
print("Skipped zero-sales days:", skipped_days)
```
