Encapsulation :

    Encapsulation is way of wrapping data(Variables and Methods) inside a capsule or a class as a Single Unit


    It restricts the direct access to some of the object components which is a means of preventing accidental inference and misuse of the data

Key Concepts of Encapsulation :

    1) Public Members
    2) Protected Members
    3) Private Members
    
    * getter and setter methods


Public Members

```python
class IronMan:

    def __init__(self,name,weapon):
        self.name = name
        self.weapon = weapon

    def intro(self):
        print(f"My name is {self.name}")
        print(f"And my weapon is {self.weapon}\n")
```

```python
tony = IronMan("Tony Stark","Mark 42")
```

dir(obj_name) --> Shows all the attributes which can be used with obj_name.

```python
dir(tony)
```

as we can see , we can directly access the attributes --> intro , name, weapon , so they are public members

```python
tony.name
```

```python
tony.__class__
```

Private Members :

    private member is not accessible outside the class and even in the derived class

    and in order to access the private members or to bring some modifications on these members , we use getter and setter methods

SYNTAX for private numbers : 

    just put __ in front of variable or attribute , while defining

```python
class Avenger:

    def __init__(self,name,ph_no):
        self.name = name
        self.__ph_no = ph_no          # here ph_no is defined as private member by adding __ at the start

    def intro(self):
        print(f"My name is {self.name}")
        print(f"My phone number is {self.ph_no}")
```

```python
stark = Avenger("Tony",5466)
stark.intro()
```

```python
dir(stark)
```

```python
stark.name
```

```python
stark.ph_no
```

as we can see , we can't access the ph_no above

To access the private member and also update the private member : 

    we use getter and setter method  --->

```python
class Animal:

    def __init__(self,name,sound):
        self.name = name
        self.__sound = sound

    def sound_info(self):
        print(f"The animal {self.name} sounds like {self.__sound}")              # getter method

    def sound_modify(self,new_sound):
        self.__sound = new_sound
        print(f"Updated Sound is {self.__sound}")                                # setter method
```

```python
cat = Animal("Cat" , "Bhow")
cat.sound_info()
```

```python
cat.sound_modify("meow")
```

Now the protected members :

    Protected members are not accessible outside of class , but it is accessible to derived class ,

    They are defined by adding one underscore in front of member _attributeName  <--- like this

```python
class Grandpa:

    def __init__(self,name,vault_pin):
        self.name = name
        self._vault_pin = vault_pin

    def pin_info(self):
        print(f"Pin for vault of {self.name} is {self._vault_pin}")
```

```python
class Parent(Grandpa):

    def __init__(self,name,vault_pin):
        super().__init__(name , vault_pin)

    def parent_reveals(self):
        print(f"Parent says the pin is : {self._vault_pin}")
```

```python
grandpa = Grandpa("Narayan",101977)
grandpa.pin_info()
```

```python
parent = Parent("Sunil",101977)
parent.parent_reveals()
```

here above , the derived class could also get access for protected data .. vault_pin
