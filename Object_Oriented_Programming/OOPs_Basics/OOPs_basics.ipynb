syntax : 

    class Class_name:

To define a constructor of class

    class Class_name :
        def __init__(self,param1,param_2,param_3,...):
            self.param1 = param1
            self.param2 = param2
            self.param3 = param3
            .
            .
            .

here self is a reference object

How to create a object and also call a constructor

    obj1 = Class_name()    # normal object is created

    obj2 = Class_name(param1,param2,param3,...)    # constructor is called

How to call a class Functin or access attributes

    print(obj1.param1)

```python
class Avenger:

    def __init__(self,name,weapon,nationality):
        self.name = name
        self.weapon = weapon
        self.nationality = nationality

    def introduction(self):
        print(f"Name of Avenger is {self.name}")
        print(f"Weapon used by {self.name} is {self.weapon}")
        print(f"Nationality is : {self.nationality}")

Iron_man = Avenger("Iron Man" , "Mark_42" , "American")

Iron_man.introduction()
```

```python
Iron_man.name = "Tony Stark"
Iron_man.name
```

```python
Iron_man.weapon
```

Creating Another object For Avenger

```python
Captain_America = Avenger("Steve Rogers" , "Shield" , "Queens")

Captain_America.introduction()
```

```python
Captain_America.weapon
```

Another Object for Avenger

```python
Thor = Avenger("Thor" , "Hammer" , "Asgard")

Thor.introduction()
```
