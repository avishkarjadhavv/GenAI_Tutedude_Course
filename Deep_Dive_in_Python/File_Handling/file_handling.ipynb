File opening Modes :


    r --> reading the content from the file
    w --> write the content into the file
    a --> append the content into the file
    r+ --> Read the content first then writing the content into the file
    w+ --> Writing the content first and then reading from it
    

To read from a file :

```python
f = open('example.txt' , 'r')
content = f.read()
print(content)
f.close()
```

**To write content into the file -
    
    * write() --> have two modes
    
        'w' --> erases the previous data , overwrites the new content
        'a' --> keep previous content and append the new content at the end**

```python
f = open('example.txt' , 'w')
f.write("This content is written with write function")
f.close()
```

To append the content into file : 

```python
f = open('example.txt' , 'a')
f.write("\n .. This line is appended in file .. ")
f.close()
```

with Keyword : It will be used for first of all opening the file as object and there is no requirement to closing the file as well

```python
with open('example.txt','r') as f:
    content = f.read()
    print(content)
```

To read or write content in binary file : 

    modes -
        'wb' --> write the content into binary file
        'rb' --> read the content in the binary file

```python
data = b'\x00\x01\x03\x04'

with open('example.txt' , 'wb') as binary_file:
    binary_file.write(data)
```

```python
with open('example.txt' , 'rb') as read_bin:
    content = read_bin.read()
    print(content)
```
