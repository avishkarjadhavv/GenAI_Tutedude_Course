Syntax for Map()

    list(map(function,iterable))

Question : Display square of all items in a list


```python
empty_list =[]

def square(list_items):
    for item in list_items:
        squared_item = item**2
        empty_list.append(squared_item)
    print(empty_list)
square([1,2,3,6,8])
```

```python
list(map(lambda x : x**2 , [1,2,3,6,8]))
```

```python
strings = [ '1', '3' , '5' , '6']

integers = list(map(int , strings))

print(strings)
print(integers)
```

```python
names = ['avishkar' , 'nilesh' , 'vinit' , 'shravan']
NAMES = list(map(str.upper , names))

print(names)
print(NAMES)
```
