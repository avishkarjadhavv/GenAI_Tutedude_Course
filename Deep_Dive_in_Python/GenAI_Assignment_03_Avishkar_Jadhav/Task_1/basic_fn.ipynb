```python
def apply_discount(price,discount_percent):
    if(discount_percent > 60):
        print("Discount exceeds 60% ...!")

    else:
        print(f"The price after applying discount of {discount_percent} percent on RS. {price} will be ", price + (price*(discount_percent/100)))

apply_discount(1000,70)
```
