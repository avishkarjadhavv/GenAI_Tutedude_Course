syntax for filter funcion 

    list(filter(function , iterable_values))

Question : Retrieve only even values from list

```python
def even_num(value):
    if(value%2 == 0):
        return True

iterable_list = [1 , 2, 4, 6 ,7, 9, 34, 46, 55 , 67 , 69]

list(filter(even_num , iterable_list))
```

```python
iterable_list = [1 , 2, 4, 6 ,7, 9, 34, 46, 55 , 67 , 69]

list(filter(lambda x : x%2 == 0 , iterable_list))
```

Question : Retrieve values less than 8

```python
iterable_list = [1 , 2, 4, 6 ,7, 9, 34, 46, 55 , 67 , 69]

list(filter(lambda x : x<8 , iterable_list))
```
